```python
import pandas as pd

df = pd.read_csv("placement_predict_50k Dataset (2).csv")

print(df.head())
print(df.columns)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.countplot(data=df, x="PlacementStatus")
plt.title("Placement Status Distribution")
plt.show()
```

```python

#BIVARIATE ANALYSIS

#Scatter Plot
plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df,
    x="CGPA",
    y="AttendancePercent"
)
plt.title("CGPA vs Attendance Percentage")
plt.show()
#Regression Plot
plt.figure(figsize=(6,4))
sns.regplot(
    data=df,
    x="CGPA",
    y="CodingTestScore"
)
plt.title("CGPA vs Coding Test Score")
plt.show()
#Box Plot
plt.figure(figsize=(6,4))
sns.boxplot(
    data=df,
    x="PlacementStatus",
    y="CGPA"
)
plt.title("CGPA vs Placement Status")
plt.show()
#Violin Plot
plt.figure(figsize=(6,4))
sns.violinplot(
    data=df,
    x="PlacementStatus",
    y="AttendancePercent"
)
plt.title("Attendance vs Placement Status")
plt.show()
#Gender vs Placement
plt.figure(figsize=(6,4))
sns.countplot(
    data=df,
    x="Gender",
    hue="PlacementStatus"
)
plt.title("Gender vs Placement Status")
plt.show()
#Stream vs Placement
plt.figure(figsize=(10,5))
sns.countplot(
    data=df,
    x="Stream",
    hue="PlacementStatus"
)
plt.xticks(rotation=45)
plt.title("Stream vs Placement Status")
plt.show()
```
